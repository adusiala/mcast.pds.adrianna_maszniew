```python
length = 56
width = 34.56
shape = "rectangle"
```

```python
print(type(length))
print(type(width))
print(type(shape))
```

```python
area = length * width
```

```python
perimeter = 2 * (length + width)
```

```python
print(f'Area of the rectangle is {area}')
print(f'Perimeter of the rectangle is {perimeter}')
```

```python
area_int = int(area)
print(area_int)
```

```python
print(type(area_int))
print(type(area))
```

```python
msg = " ".join(["The area of the", shape, "is", f"{area_int}cm2."])
```

```python
print(msg)
```
