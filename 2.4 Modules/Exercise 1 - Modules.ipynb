Date and Time Difference - Using the datetime library, write a function that takes two strings representing dates (in the format "YYYY-MM-DD") and returns the number of days between them.

```python
import datetime

def diff(date1_str, date2_str):
    date1 = datetime.datetime.strptime(date1_str, "%Y-%m-%d")
    date2 = datetime.datetime.strptime(date2_str, "%Y-%m-%d")
    return abs((date2 - date1).days)


print(diff("2021-01-01", "2025-02-16"))

```

Fetch Web Page Content - Use the urllib library to write a function that takes a URL as input and returns the HTML content of the page as a string.

```python
import urllib

def get(url):
    response = urllib.request.urlopen(url)
    content = response.read()
    return content.decode("utf-8")

print(get("https://www.google.com"))
```

Generate Random User IDs - Use the random library and create a function that generates a list of n unique user IDs, where each ID is a random integer between 1000 and 9999. Ensure no ID is repeated.

```python
import random

def random_id():
    ids = set()
    while len(ids) < 5:
        ids.add(random.randint(1000, 9999))
    return list(ids)

print(random_id())
```

Combine datetime and random - Write a script that generates a list of n random dates within a given year. Each date should be unique. Use datetime for date handling and random for the generation.

```python
import datetime
import random

def genrandates(n, year):
    start = datetime.date(year, 1, 1)
    end = datetime.date(year, 12, 31)
    num = (end - start).days + 1

    random_days = random.sample(range(num), n)
    random_dates = [start + datetime.timedelta(days=d) for d in random_days]

    return random_dates

dates = genrandates(5, 2025)
for date in dates:
    print(date)
```

Fetch and Parse JSON from an API - Use urllib to fetch data from a free-to-use API (e.g., http://jsonplaceholder.typicode.com/todos/1) and then parse this data from JSON to a Python dictionary using json.

```python
import urllib
import json

def fetch_and_parse(url):
    response = urllib.request.urlopen(url)
    
    data = response.read().decode('utf-8')
    json_data = json.loads(data)
    
    return json_data

url = "http://jsonplaceholder.typicode.com/todos/1"
result = fetch_and_parse(url)

print(result)
```

Filter Data - Add a new function makes use of the function created in question 7, so that it filters the fetched data to only include todos that are completed. Make sure to retrieve all todos by using the url https://jsonplaceholder.typicode.com/todos/.

```python
import urllib
import json

def fetch_and_parse(url):
    response = urllib.request.urlopen(url)
    
    data = response.read().decode('utf-8')
    json_data = json.loads(data)
    
    return json_data

def filter_completed_todos(url):
    todos = fetch_and_parse(url)
    
    completed_todos = [todo for todo in todos if todo['completed']]
    
    return completed_todos


url = "https://jsonplaceholder.typicode.com/todos/"
completed_todos = filter_completed_todos(url)


print(completed_todos)
```

Statistical Analysis of Random Data - Generate a list of 100 random numbers between 1 and 100 using random. Then, write a function to calculate and print the mean, median, and mode of these numbers without using any external libraries. Use math for any mathematical operations required.

```python
import random
import math

def generate_numbers(n):
    nums = []
    for i in range(n):
        nums.append(random.randint(1, 100))
    return nums


def get_stats(nums):
    mean = sum(nums) / len(nums)
    median = sorted(nums)[len(nums)//2]
    mode = max(set(nums), key=nums.count)
    return mean, median, mode


nums = generate_numbers(100)
mean, median, mode = get_stats(nums)
print (f"Mean: {mean}\nMedian: {median}\nMode: {mode}")
```
